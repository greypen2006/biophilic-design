## Melt and Concatenate DataFrames

### Subtask:
Transform `df1` and `df2` from wide to long format (year-month columns become rows), then combine them into a single DataFrame.


**Reasoning**:
The first step is to transform `df1` from a wide format to a long format. This involves identifying the ID variables and the value variables (which are the year-month columns) and then using the `melt` function to create new 'YearMonth' and 'CrimeCount' columns.



````python
id_vars = ['MajorText', 'MinorText', 'BoroughName']
value_vars_df1 = [col for col in df1.columns if col not in id_vars]
df1_melted = pd.melt(df1, id_vars=id_vars, value_vars=value_vars_df1, var_name='YearMonth', value_name='CrimeCount')
df1_melted.head()
````

**Reasoning**:
Next, melt `df2` from wide to long format using the same ID variables, creating 'YearMonth' and 'CrimeCount' columns, and then display the first few rows of the melted DataFrame.



````python
id_vars = ['MajorText', 'MinorText', 'BoroughName']
value_vars_df2 = [col for col in df2.columns if col not in id_vars]
df2_melted = pd.melt(df2, id_vars=id_vars, value_vars=value_vars_df2, var_name='YearMonth', value_name='CrimeCount')
df2_melted.head()
````

**Reasoning**:
Now that both `df1` and `df2` have been melted into a long format, the next step is to concatenate these two DataFrames (`df1_melted` and `df2_melted`) into a single DataFrame called `combined_df`.



````python
combined_df = pd.concat([df1_melted, df2_melted], ignore_index=True)
combined_df.head()
````

**Reasoning**:
To finalize the data merging process and verify the structure and data types of the newly created `combined_df`, the next step is to display a concise summary using the `.info()` method.



````python
combined_df.info()
````

## Extract and Filter by Year

### Subtask:
Extract the year from the 'YearMonth' column and filter the combined DataFrame to include only data for the years 2017 to 2024.


**Reasoning**:
The first step is to extract the year from the 'YearMonth' column in `combined_df` by converting it to a string, taking the first four characters, and then converting it to an integer type, as per the instructions.



````python
combined_df['Year'] = combined_df['YearMonth'].astype(str).str[:4].astype(int)
combined_df.head()
````

**Reasoning**:
Now that the 'Year' column has been extracted, the next step is to filter the `combined_df` to include only rows where the 'Year' is between 2017 and 2024 (inclusive) to fulfill the subtask's requirement.



````python
combined_df_filtered = combined_df[(combined_df['Year'] >= 2017) & (combined_df['Year'] <= 2024)]
combined_df_filtered.head()
````

## Filter by Boroughs

### Subtask:
Filter the DataFrame to include only the specified boroughs: Camden, Islington, Hackney, Tower Hamlets, Greenwich, Southwark, Lambeth, Hammersmith and Fulham, Kensington and Chelsea, Westminster, and Brent.


**Reasoning**:
The first step is to define the list of specified boroughs as `borough_list` to be used for filtering.



````python
borough_list = ['Camden', 'Islington', 'Hackney', 'Tower Hamlets', 'Greenwich', 'Southwark', 'Lambeth', 'Hammersmith and Fulham', 'Kensington and Chelsea', 'Westminster', 'Brent']
print("Borough list created.")
````

**Reasoning**:
Now that the `borough_list` has been defined, the next step is to filter the `combined_df_filtered` DataFrame to include only the rows where the 'BoroughName' is present in this list, as per the instructions, and store it in `combined_df_borough_filtered`.



````python
combined_df_borough_filtered = combined_df_filtered[combined_df_filtered['BoroughName'].isin(borough_list)]
combined_df_borough_filtered.head()
````
