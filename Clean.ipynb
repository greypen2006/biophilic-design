## Melt and Concatenate DataFrames

### Subtask:
Transform `df1` and `df2` from wide to long format (year-month columns become rows), then combine them into a single DataFrame.


**Reasoning**:
The first step is to transform `df1` from a wide format to a long format. This involves identifying the ID variables and the value variables (which are the year-month columns) and then using the `melt` function to create new 'YearMonth' and 'CrimeCount' columns.



````python
id_vars = ['MajorText', 'MinorText', 'BoroughName']
value_vars_df1 = [col for col in df1.columns if col not in id_vars]
df1_melted = pd.melt(df1, id_vars=id_vars, value_vars=value_vars_df1, var_name='YearMonth', value_name='CrimeCount')
df1_melted.head()
````

**Reasoning**:
Next, melt `df2` from wide to long format using the same ID variables, creating 'YearMonth' and 'CrimeCount' columns, and then display the first few rows of the melted DataFrame.



````python
id_vars = ['MajorText', 'MinorText', 'BoroughName']
value_vars_df2 = [col for col in df2.columns if col not in id_vars]
df2_melted = pd.melt(df2, id_vars=id_vars, value_vars=value_vars_df2, var_name='YearMonth', value_name='CrimeCount')
df2_melted.head()
````

**Reasoning**:
Now that both `df1` and `df2` have been melted into a long format, the next step is to concatenate these two DataFrames (`df1_melted` and `df2_melted`) into a single DataFrame called `combined_df`.



````python
combined_df = pd.concat([df1_melted, df2_melted], ignore_index=True)
combined_df.head()
````

**Reasoning**:
To finalize the data merging process and verify the structure and data types of the newly created `combined_df`, the next step is to display a concise summary using the `.info()` method.



````python
combined_df.info()
````

## Load sheet names using xlrd

### Subtask:
Load the sheet names from the Excel file `ethnic-groups-by-borough(1).xls` using `pandas.ExcelFile` with the `xlrd` engine, as `openpyxl` does not support `.xls` files. This step will identify the sheets for years 2017-2020.


**Reasoning**:
Since the previous attempt to load sheet names with `openpyxl` failed due to an unsupported file format, I will use `pandas.ExcelFile` with the `xlrd` engine to correctly read the `.xls` file and extract its sheet names, then filter them to include only the relevant years (2017-2020) as per the subtask.



````python
import pandas as pd

excel_file = pd.ExcelFile('ethnic-groups-by-borough(1).xls', engine='xlrd')
sheet_names = excel_file.sheet_names

relevant_years = [str(year) for year in range(2017, 2021)]
relevant_sheets = [sheet for sheet in sheet_names if any(year in sheet for year in relevant_years)]

print("Relevant Sheet Names:", relevant_sheets)
````

**Reasoning**:
Now that the relevant sheet names have been identified, the next step is to iterate through each of these sheets, load the data into a DataFrame, extract the year from the sheet name, add it as a 'Year' column, and then concatenate all these yearly DataFrames into a single, comprehensive DataFrame for ethnic group data.



````python
all_ethnic_dfs = []
for sheet_name in relevant_sheets:
    # Load the sheet, skipping the first 17 rows as they contain metadata
    df_year = excel_file.parse(sheet_name, skiprows=17)
    # Add a 'Year' column based on the sheet name
    df_year['Year'] = int(sheet_name) # Convert sheet name (year string) to int
    all_ethnic_dfs.append(df_year)

# Concatenate all yearly DataFrames into one
combined_ethnic_df = pd.concat(all_ethnic_dfs, ignore_index=True)

print("Combined ethnic DataFrame head:")
print(combined_ethnic_df.head())
print("\nCombined ethnic DataFrame info:")
combined_ethnic_df.info()
````
