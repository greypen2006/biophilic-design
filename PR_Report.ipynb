```python
dataframes = {
    'df_2012': df_2012,
    'df_2013': df_2013,
    'df_2014': df_2014,
    'df_2015': df_2015,
    'df_2016': df_2016,
    'df_2017': df_2017
}

for name, df_obj in dataframes.items():
    print(f"Cleaning {name}...")
    # Apply replace operations to all columns of type object
    for col in df_obj.select_dtypes(include=['object']).columns:
        df_obj[col] = df_obj[col].replace(',', '', regex=True)
        df_obj[col] = df_obj[col].replace('-', np.nan, regex=True)

    # Attempt to convert all suitable columns to float, coercing errors
    # We'll skip the first two columns (GEO_ID, NAME) as they are identifiers
    for col in df_obj.columns[2:]:
        df_obj[col] = pd.to_numeric(df_obj[col], errors='coerce')

    # Update the original dataframe in the global scope
    globals()[name] = df_obj

    print(f"Head of cleaned {name}:")
    display(df_obj.head())
    print("\n")
```
