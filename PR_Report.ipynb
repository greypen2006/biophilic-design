```python
dataframes = {
    'df_2012': df_2012,
    'df_2013': df_2013,
    'df_2014': df_2014,
    'df_2015': df_2015,
    'df_2016': df_2016,
    'df_2017': df_2017
}

# Define the target short names for the columns
# Ensure unique short names by picking one descriptive string if multiple existed
target_short_names_map = {
    'Estimate!!HOUSING OCCUPANCY!!Total housing units': 'Total housing units',
    'Margin of Error!!HOUSING OCCUPANCY!!Total housing units':'Margin of error: housing units',
    'Estimate!!HOUSING OCCUPANCY!!Vacant housing units':'Vacant housing units',
    'Margin of Error!!HOUSING OCCUPANCY!!Vacant housing units':'Margin of error: vacant units',
    'Percent!!HOUSING OCCUPANCY!!Vacant housing units':'Percentage of vacancy'
    # Removed duplicate mappings for 'Vacant housing units' and 'Percentage of vacancy'
    # 'Estimate!!HOUSING OCCUPANCY!!Total housing units!!Vacant housing units': 'Vacant housing units',
    # 'Percent!!HOUSING OCCUPANCY!!Total housing units!!Vacant housing units':'Percentage of vacancy'
}

for name, df_placeholder in dataframes.items():
    # Re-read the CSV file to ensure a consistent, raw starting state for each DataFrame
    year = int(name.split('_')[1])
    filename = f'ACSDP1Y{year}.DP04-Data.csv'
    df_obj = pd.read_csv(filename).copy()

    # 1. Store the descriptive header row (which is df_obj.iloc[0]) before dropping it.
    # This row contains the long descriptive names corresponding to DP04_XXXX codes.
    descriptive_header_values = df_obj.iloc[0]

    # 2. Drop the metadata row (row 0) from the DataFrame, which contained the descriptive headers
    df_obj = df_obj.iloc[1:].copy()

    # 3. Reset the index if it's not a default index, to ensure all desired keys are columns.
    # The 'index' column is created by reset_index. It will be dropped later if not needed.
    df_obj = df_obj.reset_index(drop=False)

    # 4. Extract year from the DataFrame name and ensure 'Year' column is correctly set.
    df_obj['Year'] = year

    # 5. Create a dynamic renaming map from DP04_XXXX codes to desired short names.
    dynamic_rename_map = {}
    for col_code in df_obj.columns:
        # Skip index-related columns for renaming from descriptive headers
        if col_code in ['GEO_ID', 'NAME', 'Year', 'index']:
            continue

        try:
            full_descriptive_name = descriptive_header_values[col_code]
            # Strip whitespace to ensure exact match with target_short_names_map keys
            full_descriptive_name_cleaned = str(full_descriptive_name).strip()
            # If this full descriptive name is one we want to process and assign a short name
            if full_descriptive_name_cleaned in target_short_names_map:
                dynamic_rename_map[col_code] = target_short_names_map[full_descriptive_name_cleaned]
        except KeyError:
            # If a column code from df_obj doesn't have a corresponding descriptive name
            # in the stored header (e.g., Unnamed columns), it's ignored for renaming.
            pass

    # 6. Rename the columns from DP04_XXXX codes to the desired short names.
    df_obj.rename(columns=dynamic_rename_map, inplace=True)

    # 7. Filter for only the desired columns (including index columns 'GEO_ID', 'NAME', 'Year')
    desired_final_columns = list(target_short_names_map.values()) + ['GEO_ID', 'NAME', 'Year']
    existing_desired_final_columns = [col for col in desired_final_columns if col in df_obj.columns]
    df_obj = df_obj[existing_desired_final_columns]

    # 8. Convert data columns to numeric (excluding index levels 'GEO_ID', 'NAME', 'Year').
    for col in df_obj.columns:
        if col not in ['GEO_ID', 'NAME', 'Year']:
            # Replace commas and hyphens before converting to numeric, if any remain
            cleaned_series = df_obj[col].astype(str).str.replace(',', '', regex=True)
            cleaned_series = cleaned_series.replace('-', np.nan, regex=True)
            df_obj[col] = pd.to_numeric(cleaned_series, errors='coerce')

    # 9. Set the desired multi-index.
    df_obj.set_index(['GEO_ID', 'NAME', 'Year'], inplace=True)

    # 10. Update the original dataframe in the global scope
    globals()[name] = df_obj

    print(f"Head of {name} with new index and renamed columns (showing only processed columns):")
    # Display only the renamed columns that are now in df_obj.columns
    display(df_obj[list(target_short_names_map.values())].head())
    print("\n")
```

```python
# Read the CSV without a header initially to capture all necessary rows for column names
df_raw = pd.read_csv('PuertoRico (2).csv', header=None)

# Extract the first two rows which contain the information for the multi-level column headers
header0_orig = df_raw.iloc[0].copy() # Contains 'Unnamed: 0', 'Puerto Rico', 'Unnamed: 2', ...
header1_orig = df_raw.iloc[1].copy() # Contains 'Year', 'Total Housing Units', 'Margin of error', ...

# Construct the top-level header (geographic areas) based on original column names and user request
final_level0 = [''] * len(df_raw.columns)
final_level0[0] = 'Year'

# Dynamically assign geographic names based on header0_orig
current_geo = None
for col_idx, geo_name in header0_orig.items():
    if col_idx == 0: # Skip the 'Year' column which is handled separately
        continue

    if pd.notna(geo_name): # If a geographic name is present in header0_orig, update current_geo
        current_geo = str(geo_name).strip()

    # Assign the current geographic name to the corresponding column in final_level0
    if current_geo and col_idx < len(final_level0): # Ensure we don't go out of bounds
        final_level0[col_idx] = current_geo

# Clean the second-level header (metric names)
# Remove leading/trailing spaces and extra spaces within names like 'Rent  between 500-750'
level1_header_cleaned = [str(x).strip().replace('  ', ' ') if pd.notna(x) else x for x in header1_orig.tolist()]

# Create the MultiIndex for the DataFrame's columns
multi_index = pd.MultiIndex.from_arrays([final_level0, level1_header_cleaned])

# Assign this MultiIndex to the DataFrame's data, skipping the first two header rows
vacancy = df_raw.iloc[2:].copy()
vacancy.columns = multi_index

# Set the 'Year' column (which is now ('Year', 'Year')) as the DataFrame's index
vacancy = vacancy.set_index(('Year', 'Year'))
vacancy.index = pd.to_numeric(vacancy.index) # Convert the year index to numeric
vacancy.index.rename('Year', inplace=True) # Rename the index level for clarity

vacancy.head()
```

```python
columns_to_convert = [
    ('Puerto Rico', 'Percentage of Vacancy PR'),
    ('Puerto Rico', 'Population PR'),
    ('Puerto Rico', 'Median Income PR'),
    ('Puerto Rico', 'Rent less than 500 PR'),
    ('San Juan', 'Percentage of Vacancy'),
    ('San Juan', 'Rent less than 500'),
    ('San Juan', 'Population'),
    ('San Juan', 'Median Income')
]

for col_tuple in columns_to_convert:
    if col_tuple in vacancy.columns:
        # Convert to numeric, coercing errors to NaN
        vacancy[col_tuple] = pd.to_numeric(vacancy[col_tuple], errors='coerce')
    else:
        print(f"Warning: Column {col_tuple} not found in vacancy DataFrame. Skipping conversion.")

print("Data types after conversion:")
display(vacancy.info())
print("\nHead of vacancy DataFrame after numeric conversion:")
display(vacancy.head())
```

```python
vacant='Percentage of Vacancy'
income='Median Income'
population='Population'
affordability="Rent less than 500"
```

```python
models=[] # create empty list to store the models
names=[] # create empty list to store the names of the models

# Get unique geographic areas from the top level of the column MultiIndex
geo_areas = vacancy.columns.get_level_values(0).unique()

# These are the *generic* variable names we want to model
generic_vacant = 'Percentage of Vacancy'
generic_income = 'Median Income'
generic_population = 'Population'
generic_affordability = 'Rent less than 500'

# Define simplified names for OLS formula construction from generic terms
ols_formula_vars = {
    generic_vacant: 'vacant_rate',
    generic_income: 'median_income',
    generic_population: 'population_size',
    generic_affordability: 'rent_affordability'
}

for geo_area in geo_areas:
    # Initialize a temporary DataFrame to hold the processed data for OLS
    # The index will be the year index from the original vacancy DataFrame
    temp_df_for_ols = pd.DataFrame(index=vacancy.index)

    # Map generic variable names to the *actual* column names in geo_df for this specific area
    # Adjusted to check for 'Puerto Rico' as it is the current top-level name
    if geo_area == 'Puerto Rico':
        actual_vacant_col = 'Percentage of Vacancy PR'
        actual_income_col = 'Median Income PR'
        actual_population_col = 'Population PR'
        actual_affordability_col = 'Rent less than 500 PR'
    elif geo_area == 'San Juan':
        actual_vacant_col = 'Percentage of Vacancy'
        actual_income_col = 'Median Income'
        actual_population_col = 'Population'
        actual_affordability_col = 'Rent less than 500'
    else:
        print(f"Skipping unknown geographic area: {geo_area}")
        continue

    # Explicitly extract columns one by one and assign to temp_df_for_ols with simplified names
    # This prevents issues with pandas slicing returning DataFrames with duplicate column labels
    try:
        temp_df_for_ols[ols_formula_vars[generic_vacant]] = vacancy[(geo_area, actual_vacant_col)]
        temp_df_for_ols[ols_formula_vars[generic_income]] = vacancy[(geo_area, actual_income_col)]
        temp_df_for_ols[ols_formula_vars[generic_population]] = vacancy[(geo_area, actual_population_col)]
        temp_df_for_ols[ols_formula_vars[generic_affordability]] = vacancy[(geo_area, actual_affordability_col)]
    except KeyError as e:
        print(f"Skipping regression for {geo_area}: Missing expected column - {e}")
        continue

    # Drop rows with NaN values before regression
    temp_df_for_ols.dropna(inplace=True)

    if not temp_df_for_ols.empty:
        try:
            # Use the simplified, consistent names in the OLS formula
            model = ols(f'{ols_formula_vars[generic_vacant]} ~ {ols_formula_vars[generic_income]} + {ols_formula_vars[generic_population]} + {ols_formula_vars[generic_affordability]}', data=temp_df_for_ols).fit()
            models.append(model)
            names.append(f'{geo_area}_Overall')
        except Exception as e:
            print(f"Error running regression for {geo_area}: {e}")
    else:
        print(f"Skipping regression for {geo_area}: Not enough valid data after dropping NaNs.")

# Create a regression table if models were generated
if models:
    table = summary_col(models, stars=True, float_format='%0.3f', model_names=names, info_dict = {"N":lambda x: "{0:d}".format(int(x.nobs))}) # add the number of observations to the table
    print(table)
else:
    print("No models were generated. Check data and column names.")
```

```python
from linearmodels import PanelOLS
from linearmodels import RandomEffects
import statsmodels.formula.api as smf
from linearmodels.panel import compare
import pandas as pd # Ensure pandas is imported if not already globally available

# Reuse definitions from vmkxmlaHn7fA to ensure consistency
generic_vacant = 'Percentage of Vacancy'
generic_income = 'Median Income'
generic_population = 'Population'
generic_affordability = 'Rent less than 500'

ols_formula_vars = {
    generic_vacant: 'vacant_rate',
    generic_income: 'median_income',
    generic_population: 'population_size',
    generic_affordability: 'rent_affordability'
}

# Create an empty list to store processed dataframes for panel data
panel_dataframes = []

# Get unique geographic areas from the top level of the column MultiIndex
geo_areas = vacancy.columns.get_level_values(0).unique()

for geo_area in geo_areas:
    # Initialize a temporary DataFrame to hold the processed data for OLS
    # The index will be the year index from the original vacancy DataFrame
    temp_df_for_panel = pd.DataFrame(index=vacancy.index)

    # Map generic variable names to the *actual* column names in vacancy for this specific area
    if geo_area == 'Puerto Rico': # Corrected: Check for 'Puerto Rico'
        actual_vacant_col = 'Percentage of Vacancy PR'
        actual_income_col = 'Median Income PR'
        actual_population_col = 'Population PR'
        actual_affordability_col = 'Rent less than 500 PR'
        entity_id = 'Puerto Rico' # Identifier for this entity
    elif geo_area == 'San Juan':
        actual_vacant_col = 'Percentage of Vacancy'
        actual_income_col = 'Median Income'
        actual_population_col = 'Population'
        actual_affordability_col = 'Rent less than 500'
        entity_id = 'San Juan' # Identifier for this entity
    else:
        print(f"Skipping unknown geographic area: {geo_area} for panel data creation.")
        continue

    # Explicitly extract columns from 'vacancy' using full MultiIndex tuple
    # and assign to temp_df_for_panel with simplified names
    try:
        temp_df_for_panel[ols_formula_vars[generic_vacant]] = vacancy[(geo_area, actual_vacant_col)]
        temp_df_for_panel[ols_formula_vars[generic_income]] = vacancy[(geo_area, actual_income_col)]
        temp_df_for_panel[ols_formula_vars[generic_population]] = vacancy[(geo_area, actual_population_col)]
        temp_df_for_panel[ols_formula_vars[generic_affordability]] = vacancy[(geo_area, actual_affordability_col)]
    except KeyError as e:
        print(f"Skipping entity {entity_id} for panel data creation due to missing column: {e}")
        continue

    # Add the entity_id as a column
    temp_df_for_panel['geo_area'] = entity_id
    panel_dataframes.append(temp_df_for_panel)

# Concatenate all entity-specific dataframes into one for PanelOLS
df_c = pd.concat(panel_dataframes)

# Drop rows with any NaN values that might prevent panel regression
df_c.dropna(inplace=True)

# Set the MultiIndex for PanelOLS: (entity_id, time_id)
# The current index of temp_df_for_panel is 'Year', so df_c.index is 'Year'.
# We need to set 'geo_area' as the first level and 'Year' (current index) as the second.
df_c = df_c.set_index('geo_area', append=True) # Makes index (Year, geo_area)
df_c = df_c.swaplevel(0, 1) # Makes index (geo_area, Year)
df_c.index.names = ['geo_area', 'year'] # Rename index levels

# Now df_c is in the correct format for PanelOLS

# Run a fixed effects model using the simplified variable names
panel = PanelOLS.from_formula(
    f'{ols_formula_vars[generic_vacant]} ~ 1 + {ols_formula_vars[generic_population]} + {ols_formula_vars[generic_income]} + {ols_formula_vars[generic_affordability]} + EntityEffects',
    data=df_c
).fit()

# Print the model formatted as a regression table
print(compare({'Fixed Effects': panel,}, stars=True))
```
